```python
import numpy as np
```

```python
image_matrix = np.array([
    [100, 150, 200, 250],
    [50, 120, 180, 220],
    [30, 90, 140, 190],
    [10, 50, 100, 150]
])

darkened_image = np.clip(image_matrix - 20, 0, 255)
mean_intensity = darkened_image.mean()

print("Darkened image:")
print(darkened_image)
print(f"Mean intensity: {mean_intensity:.2f}")
```

```python
coordinates = np.array([[1.5, 2.0], [5.0, 8.1], [3.2, 1.1], [8.9, 4.3], [2.1, 2.8]])

x_coordinates, y_coordinates = coordinates.T
max_x = x_coordinates.max()
min_y = y_coordinates.min()
selected_coordinates = coordinates[(x_coordinates + y_coordinates) > 6.0]

print("X coordinates:", x_coordinates)
print("Y coordinates:", y_coordinates)
print("Maximum X:", max_x)
print("Minimum Y:", min_y)
print("Points where X + Y > 6.0:")
print(selected_coordinates)
```

```python
stock_prices = np.array([100.0, 102.5, 101.0, 105.0, 104.2, 108.0, 110.5])

today_prices = stock_prices[1:]
yesterday_prices = stock_prices[:-1]
daily_returns = (today_prices - yesterday_prices) / yesterday_prices
best_return_index = np.argmax(daily_returns)
best_day = best_return_index + 2

print("Today's prices:", today_prices)
print("Yesterday's prices:", yesterday_prices)
print("Daily returns (%):", daily_returns * 100)
print(f"Highest return: day {best_day}, {daily_returns[best_return_index] * 100:.2f}%")
```

```python
sensor_data = np.array([23.4, 24.1, -999, 25.0, 26.2, -999, 24.8])

missing_mask = sensor_data == -999
valid_mean = sensor_data[~missing_mask].mean()
filled_sensor_data = np.where(missing_mask, valid_mean, sensor_data)

print("Missing-value mask:", missing_mask)
print(f"Mean of valid values: {valid_mean:.2f}")
print("Data after filling:", filled_sensor_data)
```

```python
inventory = np.array([12, 45, 23, 67, 89, 34, 50])
target = 40

differences = np.abs(inventory - target)
closest_index = np.argmin(differences)
closest_value = inventory[closest_index]

print("Absolute differences:", differences)
print("Index of closest value:", closest_index)
print("Closest value:", closest_value)
```
