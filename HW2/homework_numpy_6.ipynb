```python
import numpy as np
import pandas as pd

first = np.arange(1, 6)
second = np.arange(6, 11)

combined = np.concatenate((first, second))
array_2x5 = np.vstack((first, second))
array_5x2 = np.column_stack((first, second))

print("Об'єднаний одновимірний масив:")
display(pd.DataFrame({"Значення": combined}))
print("Двовимірний масив 2 x 5:")
display(pd.DataFrame(array_2x5))
print("Двовимірний масив 5 x 2:")
display(pd.DataFrame(array_5x2))
```
