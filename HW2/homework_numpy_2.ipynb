```python
import numpy as np
import pandas as pd

array = np.random.randint(1, 51, size=(4, 4))

display(pd.DataFrame(array, index=["Рядок 1", "Рядок 2", "Рядок 3", "Рядок 4"],
                     columns=["Стовпець 1", "Стовпець 2", "Стовпець 3", "Стовпець 4"]))

row_sums = np.sum(array, axis=1)
column_sums = np.sum(array, axis=0)

display(pd.DataFrame({"Рядок": [1, 2, 3, 4], "Сума": row_sums}))
display(pd.DataFrame({"Стовпець": [1, 2, 3, 4], "Сума": column_sums}))
print("Середнє значення масиву:", np.mean(array))
```
