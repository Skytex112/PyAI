```python
import numpy as np
import pandas as pd

numbers = np.random.randint(-10, 11, size=15)
positive_numbers = numbers[numbers > 0]
even_numbers = numbers[numbers % 2 == 0]

display(pd.DataFrame({
    "Початковий масив": pd.Series(numbers),
    "Додатні числа": pd.Series(positive_numbers),
    "Парні числа": pd.Series(even_numbers),
}))
```
