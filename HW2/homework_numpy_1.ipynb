```python
import numpy as np
import pandas as pd

numbers = np.random.randint(0, 101, size=10)
sorted_numbers = np.sort(numbers)

display(pd.DataFrame({
    "До сортування": numbers,
    "Після сортування": sorted_numbers,
}))

display(pd.DataFrame({
    "Показник": ["Мінімум", "Максимум", "Середнє"],
    "Значення": [np.min(numbers), np.max(numbers), np.mean(numbers)],
}))
```
