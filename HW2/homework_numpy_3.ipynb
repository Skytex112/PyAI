```python
import numpy as np
import pandas as pd

array = np.arange(15).reshape(3, 5)
transposed = array.T
reshaped = array.reshape(5, 3)

print("Вихідний масив 3 x 5:")
display(pd.DataFrame(array))
print("Транспонований масив:")
display(pd.DataFrame(transposed))
print("Масив після reshape(5, 3):")
display(pd.DataFrame(reshaped))
```
