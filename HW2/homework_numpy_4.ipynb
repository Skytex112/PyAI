```python
import numpy as np
import pandas as pd

array = np.random.random((10, 10))
min_index = np.unravel_index(np.argmin(array), array.shape)
max_index = np.unravel_index(np.argmax(array), array.shape)

display(pd.DataFrame(array))
display(pd.DataFrame({
    "Елемент": ["Мінімальний", "Максимальний"],
    "Значення": [np.min(array), np.max(array)],
    "Індекс (рядок, стовпець)": [min_index, max_index],
}))
```
