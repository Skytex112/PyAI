```python
import pandas as pd
from IPython.display import display

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

results = pd.DataFrame({
    "Середній бал": [sum(math) / len(math), sum(physics) / len(physics), sum(english) / len(english)],
    "Мінімальний бал": [min(math), min(physics), min(english)],
    "Максимальний бал": [max(math), max(physics), max(english)],
}, index=["Math", "Physics", "English"])

display(results)
print("Предмет із найвищим середнім балом:", results["Середній бал"].idxmax())
```
