```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

population = np.random.exponential(scale=2, size=50000)
print("Population size:", len(population))

number_of_samples = 1000
samples_5 = np.random.choice(population, size=(number_of_samples, 5))
samples_30 = np.random.choice(population, size=(number_of_samples, 30))
means_5 = samples_5.mean(axis=1)
means_30 = samples_30.mean(axis=1)

df_means = pd.DataFrame({"mean_n_5": means_5, "mean_n_30": means_30})
display(df_means.head())

print(f"n = 5: mean = {means_5.mean():.2f}, standard deviation = {means_5.std():.2f}")
print(f"n = 30: mean = {means_30.mean():.2f}, standard deviation = {means_30.std():.2f}")

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.hist(means_5, bins=30, edgecolor="black")
plt.title("Sample means, n = 5")
plt.xlabel("Sample mean")

plt.subplot(1, 2, 2)
plt.hist(means_30, bins=30, edgecolor="black")
plt.title("Sample means, n = 30")
plt.xlabel("Sample mean")
plt.tight_layout()
plt.show()
```
