```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

dates = pd.date_range("2026-01-01", periods=90)
day = np.arange(90)
w = 0.8
b = 120
sales = w * day + b + np.random.normal(0, 12, 90)

df = pd.DataFrame({"date": dates, "sales": sales})

df["moving_average"] = df["sales"].rolling(7).mean()
df["moving_std"] = df["sales"].rolling(7).std()
display(df.head(10))

plt.plot(df["date"], df["sales"], label="Sales")
plt.plot(df["date"], df["moving_average"], label="Moving average")
plt.title("Sales and moving average")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.plot(df["date"], df["moving_std"], color="green")
plt.title("Moving standard deviation")
plt.xlabel("Date")
plt.ylabel("Standard deviation")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
