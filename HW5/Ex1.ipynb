```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

dates = pd.date_range("2026-01-01", periods=30)
users = 500 + np.arange(30) * 5 + np.random.randint(-30, 30, 30)
sessions = users * 2 + np.random.randint(-50, 50, 30)
revenue = sessions * 5 + np.random.randint(-100, 100, 30)

df = pd.DataFrame({"date": dates, "users": users, "sessions": sessions, "revenue": revenue})
display(df)

display(df[["users", "sessions", "revenue"]].corr())

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.scatter(df["users"], df["sessions"])
plt.xlabel("Users")
plt.ylabel("Sessions")

plt.subplot(1, 3, 2)
plt.scatter(df["users"], df["revenue"])
plt.xlabel("Users")
plt.ylabel("Revenue")

plt.subplot(1, 3, 3)
plt.scatter(df["sessions"], df["revenue"])
plt.xlabel("Sessions")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()

plt.plot(df["date"], df["revenue"])
plt.title("Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
