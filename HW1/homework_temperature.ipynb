```python
import pandas as pd

temperatures = [2, 5, 4, 6, 7, 3, 1]

average = sum(temperatures) / len(temperatures)
maximum = max(temperatures)
minimum = min(temperatures)
difference = maximum - minimum

results = pd.DataFrame({
    "Показник": ["Середнє значення", "Максимальне значення", "Мінімальне значення", "Різниця між максимумом і мінімумом"],
    "Значення": [average, maximum, minimum, difference],
})
display(results)
```
