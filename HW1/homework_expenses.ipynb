```python
import pandas as pd

expenses = [1200, 1500, 800, 600, 1000, 400, 900]

total = sum(expenses)
average = total / len(expenses)
maximum = max(expenses)
minimum = min(expenses)
days_above_1000 = sum(expense > 1000 for expense in expenses)

results = pd.DataFrame({
    "Показник": ["Загальна сума витрат", "Середнє значення", "Максимальна витрата", "Мінімальна витрата", "Днів із витратами понад 1000"],
    "Значення": [total, round(average, 2), maximum, minimum, days_above_1000],
})
display(results)
```
