```python
from collections import Counter
import pandas as pd

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
vote_counts = Counter(votes)

leader, leader_votes = vote_counts.most_common(1)[0]
top_3 = vote_counts.most_common(3)

display(pd.DataFrame(vote_counts.items(), columns=["Фрукт", "Кількість голосів"]))
print("Топ-3:")
display(pd.DataFrame(top_3, columns=["Фрукт", "Кількість голосів"]))
display(pd.DataFrame({"Лідер": [leader], "Кількість голосів": [leader_votes]}))
```
