```python
import matplotlib.pyplot as plt

hours = list(range(24))
load = [22, 18, 15, 13, 12, 16, 25, 38, 55, 63, 70, 74,
        68, 65, 72, 80, 88, 95, 91, 78, 64, 52, 40, 30]

plt.plot(hours, load, color="tab:blue", marker="o", markersize=4)
plt.fill_between(hours, load, color="tab:blue", alpha=0.25)
plt.xlabel("Година")
plt.ylabel("Навантаження сервера, %")
plt.title("Погодинне навантаження сервера")
plt.xticks(hours)
plt.grid(True, alpha=0.3)
plt.show()
```
