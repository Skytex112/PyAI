```python
import matplotlib.pyplot as plt

weeks = ["Тиждень 1", "Тиждень 2", "Тиждень 3", "Тиждень 4", "Тиждень 5", "Тиждень 6"]
conversion = [3.2, 3.5, 3.4, 3.8, 4.0, 4.2]
retention = [42, 44, 43, 47, 49, 51]
average_order = [820, 850, 830, 900, 940, 960]
orders = [1200, 1280, 1250, 1390, 1480, 1560]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].plot(weeks, conversion, marker="o")
axes[0, 0].set_title("Конверсія, %")
axes[0, 1].plot(weeks, retention, marker="o", color="tab:orange")
axes[0, 1].set_title("Утримання, %")
axes[1, 0].plot(weeks, average_order, marker="o", color="tab:green")
axes[1, 0].set_title("Середній чек, грн")
axes[1, 1].plot(weeks, orders, marker="o", color="tab:red")
axes[1, 1].set_title("Кількість замовлень")

for axis in axes.flat:
    axis.set_xlabel("Період")
    axis.grid(True, alpha=0.3)
    axis.tick_params(axis="x", rotation=30)

fig.suptitle("Динаміка метрик продукту", fontsize=16)
fig.tight_layout()
plt.show()
```
