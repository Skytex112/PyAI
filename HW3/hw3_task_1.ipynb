```python
import matplotlib.pyplot as plt

months = ["Січень", "Лютий", "Березень", "Квітень", "Травень", "Червень"]
plan = [120, 135, 150, 145, 165, 180]
actual = [115, 140, 142, 155, 160, 190]

plt.plot(months, plan, marker="o", label="План")
plt.plot(months, actual, marker="o", label="Факт")
plt.xlabel("Місяць")
plt.ylabel("Продажі, тис. грн")
plt.title("Планові та фактичні продажі")
plt.legend()
plt.grid(True)
plt.show()
```
