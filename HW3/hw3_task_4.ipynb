```python
import matplotlib.pyplot as plt

days = ["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Нд"]
temperature = [18, 20, 19, 22, 24, 23, 21]
humidity = [65, 60, 68, 55, 50, 53, 59]

plt.plot(days, temperature, marker="o", label="Температура, °C")
plt.plot(days, humidity, marker="o", label="Вологість, %")
plt.xlabel("День тижня")
plt.ylabel("Значення")
plt.title("Температура та вологість за тиждень")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
